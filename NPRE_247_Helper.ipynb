```python
import pandas as pd
import sympy as sp
import os 
import sys
```

```python
#Load NPRE_247_Data
def resource_path(relative_path):
    try:
        # PyInstaller creates a temp folder and stores path in _MEIPASS
        base_path = sys._MEIPASS
    except Exception:
        base_path = os.path.abspath(".")

    return os.path.join(base_path, relative_path)
    
file_path = resource_path("NPRE_247_Data.txt")
```

```python
#Setup
N = 1
df = pd.read_csv(file_path)
df.columns = ["N","Z","A","EL","Atomic Mass (u)"]
df["A"] = df["N"] + df["Z"]
#df.iloc[N:, df.columns.get_loc("A")] = df.iloc[N:, df.columns.get_loc("N")] + df.iloc[N:, df.columns.get_loc("Z")]
```

```python
def welcome():
    print("Welcome to the UIUC NPRE 247 Class Helper")
    print("This program is intended to help with repeated calculations found in the NPRE 247 course")
    print("Btw 1 is always used to quit")
```

```python
#Q_value calclator mode
def Q_value():
    quit = False
    
    while quit != True:
        print("")
        print("Q-value Calulator")
        reatants = input("How many rectants:")
        rea_array = sp.Matrix.zeros(1,int(reatants))

        products = input("How many products")
        pro_array = sp.Matrix.zeros(1,int(products))

        print("")
        print("Notation Ex. Oxygen 16 = O-16")
        for i in range(len(rea_array)):
            reactant_input = input(f"What is the {i+1} reactnat")

            if len(reactant_input) == 1:
                rea_array[i] = float(df.loc[(df["EL"] == reactant_input), "Atomic Mass (u)"].item())
                
            else:
                result = reactant_input.split("-")
                rea_array[i] = float(df.loc[(df["EL"] == result[0]) & (df["A"] == int(result[1])), "Atomic Mass (u)"].item())
    
        for i in range(len(pro_array)):
            product_input = input(f"What is the {i+1} product")

            if len(product_input) == 1:
                pro_array[i] = float(df.loc[(df["EL"] == product_input), "Atomic Mass (u)"].item())

            else:
                result_p = product_input.split("-")
                pro_array[i] = float(df.loc[(df["EL"] == result_p[0]) & (df["A"] == int(result_p[1])), "Atomic Mass (u)"].item())
    

        print("")
        print("Q_value is =")
        print(f"{rea_array} - {pro_array} x 931.5")
        mass_change = (sum(rea_array) - sum(pro_array)) * 931.5
        print(f"{mass_change} MeV")

        print("")
        stop = input("Press enter for more Q or 1 to quit")
        if stop == "1":
            break
```

```python
#Main Enginge loop

close = False
program_names = ["Q Value Calulator"]
programs = [Q_value]
alp = ["A","B","C","D","E","F","G"]
start = True

while close == False:
    if start == True:
        welcome()
        start = False
    
    print("")
    print("The programs included are:")

    for i in range(len(program_names)):
        print(f"Program {alp[i]}: {program_names[i]}")
    
    print("")
    select = input("Now what program would you like to use(Ex = A)?")

    if select == "1":
        break

    else:
        programs[(alp.index(select))]()
    
    
```
